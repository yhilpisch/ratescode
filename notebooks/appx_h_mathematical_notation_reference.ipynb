## Mini-lab: Translate a credit formula into numbers

The mini-lab evaluates one notation-heavy expression from the credit chapters with every symbol stated: expected loss equals the probability of default times loss given default times exposure at default. Default is driven by a constant hazard rate (default intensity) of 3% per year over a two-year horizon, so the survival probability is the exponential of minus 0.03 times 2. The recovery rate is 40% of exposure, loss given default is 1 minus recovery, and the exposure at default is 10 million dollars. All inputs are illustrative.

```python
import numpy as np
import pandas as pd

intensity = 0.03  # default intensity, constant hazard rate per year
horizon = 2.0  # time horizon in years
recovery = 0.40  # recovery rate on defaulted exposure
ead = 10_000_000  # exposure at default in dollars
survival = np.exp(-intensity * horizon)  # survival probability
default_prob = 1 - survival  # probability of default by the horizon
lgd = 1 - recovery  # loss given default as a fraction
expected_loss = default_prob * lgd * ead  # expected loss in dollars
pd.Series({"survival": survival, "default_prob": default_prob,
           "lgd": lgd, "expected_loss": expected_loss})
```

### What to observe

The survival probability is 0.941765, so the probability of default within two years is 0.058235, or 5.82%. With loss given default of 0.60, the expected loss is 0.058235 times 0.60 times 10,000,000, or about 349,413 dollars. Each symbol in the chain carries a stated unit: a probability, a fraction, and a dollar exposure multiply to a dollar loss.
