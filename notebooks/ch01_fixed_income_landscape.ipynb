## Mini-lab: Map instruments to cash flows and risk drivers

The chapter reduces the fixed-income landscape to one testable object: a table that pairs each instrument with its cash-flow pattern and its dominant risk driver. The rows below mirror the chapter's instrument map; once the map is a DataFrame, Python can filter and aggregate it like any other data structure. All entries are illustrative inputs, not market data.

```python
import pandas as pd

instruments = pd.DataFrame([
    {"instrument": "Treasury bill", "cashflows": "single payment",
     "dominant_risk": "roll-down"},
    {"instrument": "Fixed-rate bond", "cashflows": "coupons",
     "dominant_risk": "duration"},
    {"instrument": "Swap", "cashflows": "two legs",
     "dominant_risk": "curve shape"},
    {"instrument": "Corporate bond", "cashflows": "coupons",
     "dominant_risk": "spread"},
])
risk_counts = instruments.groupby("dominant_risk").size()
print(risk_counts.sort_index().to_string())
instruments
```

### What to observe

With these inputs the calculation yields four instrument rows and a count of one instrument per dominant risk driver: roll-down, duration, curve shape, and spread. The point is structural rather than numerical: once instruments, cash-flow patterns, and risk drivers live in a DataFrame, the same map supports the filtering, aggregation, pricing, and risk workflows later chapters build on.
